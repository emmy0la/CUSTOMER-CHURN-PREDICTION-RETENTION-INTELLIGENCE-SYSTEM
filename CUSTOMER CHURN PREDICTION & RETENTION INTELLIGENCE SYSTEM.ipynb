```python
# ============================================================
# 2.5 COUNT CHURNED CUSTOMERS
# ============================================================

# Now I want to count customers where:
# Churn = 1
#
# Churn = 1 means the customer left.
# Churn = 0 means the customer stayed.
#
# I use a loop to count all churned customers.
# ============================================================

print("=" * 60)
print("2.5 COUNT CHURNED CUSTOMERS")
print("=" * 60)
print()

churned_customers = 0

for customer in customers:
    if customer["Churn"] == 1:
        churned_customers += 1

print("Number of Churned Customers:", churned_customers)
print()

# ---- Also count retained customers ----

retained_customers = 0

for customer in customers:
    if customer["Churn"] == 0:
        retained_customers += 1

print("Number of Retained Customers:", retained_customers)
print()

# ---- Verify ----

print("Verification:")
print("Churned + Retained =", churned_customers + retained_customers)
print("Total Customers =", total_customers)
print()

if churned_customers + retained_customers == total_customers:
    print("Status: PASSED - All customers accounted for")
else:
    print("Status: REVIEW REQUIRED")
print()
```

```python
# ============================================================
# 2.6 CALCULATE CHURN RATE
# ============================================================

# Churn Rate formula:
#
# Churn Rate = (Churned Customers / Total Customers) × 100
#
# This tells me the percentage of customers who left.
# ============================================================

print("=" * 60)
print("2.6 CALCULATE CHURN RATE")
print("=" * 60)
print()

churn_rate = (churned_customers / total_customers) * 100

print("Churn Rate Calculation:")
print("-" * 40)
print(f"Churned Customers: {churned_customers}")
print(f"Total Customers: {total_customers}")
print()
print(f"Churn Rate = ({churned_customers} / {total_customers}) × 100")
print(f"Churn Rate = {churn_rate:.2f}%")
print()
```

```python
# ============================================================
# 3.1 IMPORT REQUIRED LIBRARY
# ============================================================

# I am using Python's built-in "random" module.
#
# Why "random"?
# - It is a Python standard library (no external installation needed)
# - It allows me to generate realistic synthetic data
# - It keeps the project within the scope of what I have learned
#
# I am NOT using:
# - pandas
# - numpy
# - any external library
#
# This keeps the project focused on Python fundamentals.
# ============================================================

import random

print("=" * 60)
print("SECTION 3: SCALING TO 10,000 CUSTOMERS")
print("=" * 60)
print()
print("Step 1: Importing the random module")
print("-" * 60)
print()
print("random module imported successfully.")
print("This is a Python standard library.")
print("No external installation is required.")
print()
```

```python
# ============================================================
# 3.3 DEFINE CATEGORIES
# ============================================================

# Before generating the dataset, I define the categories
# for all categorical variables.
#
# This ensures consistency across all 10,000 records.
#
# I am using realistic categories for a Nigerian telecom company.
# ============================================================

print("=" * 60)
print("Step 3: Defining categories for categorical variables")
print("-" * 60)
print()

# ---- Genders ----
genders = ["Male", "Female"]

# ---- Regions ----
# Nigerian states representing different geographic areas
regions = [
    "Lagos",
    "Abuja",
    "Rivers",
    "Oyo",
    "Kano",
    "Enugu",
    "Kaduna",
    "Ogun"
]

# ---- Subscription Types ----
subscription_types = [
    "Basic",
    "Standard",
    "Premium"
]

# ---- Contract Types ----
contract_types = [
    "Monthly",
    "Quarterly",
    "Annual"
]

# ---- Payment Methods ----
payment_methods = [
    "Bank Transfer",
    "Card",
    "USSD",
    "Mobile Money"
]

# ---- Device Types ----
device_types = [
    "Android",
    "iPhone",
    "Feature Phone"
]

# ---- Display all categories ----

print("Genders:", genders)
print()
print("Regions:", regions)
print()
print("Subscription Types:", subscription_types)
print()
print("Contract Types:", contract_types)
print()
print("Payment Methods:", payment_methods)
print()
print("Device Types:", device_types)
print()
print("Categories defined successfully.")
print()
```

```python
# ============================================================
# 3.4 GENERATE 10,000 CUSTOMER RECORDS
# ============================================================

# Now I generate the full dataset.
#
# I am using a "for" loop to create 10,000 records.
#
# Each record is a dictionary with 23 fields.
#
# For numeric fields, I use random.randint() or random.uniform()
# to generate realistic values.
#
# For categorical fields, I use random.choice() to select
# from the categories I defined.
# ============================================================

print("=" * 60)
print("Step 4: Generating 10,000 customer records")
print("-" * 60)
print()

customers = []

for number in range(1, number_of_customers + 1):

    # ---- Generate independent variables ----
    satisfaction = random.randint(1, 5)
    complaints = random.randint(0, 6)
    support_calls = random.randint(0, 10)
    network_issues = random.randint(0, 8)
    last_activity = random.randint(0, 60)

    # ---- Generate churn BASED ON these variables ----
    # This is what creates signal in the data

    if satisfaction <= 2 and complaints >= 3 and last_activity >= 15:
        # HIGH RISK: Strong signals
        churn = 1 if random.random() < 0.85 else 0

    elif satisfaction <= 3 or complaints >= 2 or last_activity >= 10:
        # MEDIUM RISK: Moderate signals
        churn = 1 if random.random() < 0.50 else 0

    else:
        # LOW RISK: Weak signals
        churn = 1 if random.random() < 0.10 else 0

    customer = {
        "Customer_ID": f"C{number:05d}",
        "Age": random.randint(18, 70),
        "Gender": random.choice(genders),
        "Region": random.choice(regions),
        "Tenure_Months": random.randint(1, 72),
        "Subscription_Type": random.choice(subscription_types),
        "Contract_Type": random.choice(contract_types),
        "Monthly_Charges": random.randint(2000, 30000),
        "Total_Charges": random.randint(2000, 1500000),
        "Data_Usage_GB": round(random.uniform(0.5, 100), 2),
        "Voice_Minutes": random.randint(0, 5000),
        "SMS_Count": random.randint(0, 2000),
        "Payment_Method": random.choice(payment_methods),
        "Payment_Delay_Count": random.randint(0, 8),
        "Complaints_Count": complaints,
        "Support_Calls": support_calls,
        "Network_Issues": network_issues,
        "Service_Satisfaction": satisfaction,
        "Device_Type": random.choice(device_types),
        "Auto_Renewal": random.choice([0, 1]),
        "Discount_Received": random.choice([0, 1]),
        "Last_Activity_Days": last_activity,
        "Churn": churn
    }

    customers.append(customer)

print("Dataset created successfully.")
print("Total customer records:", len(customers))
print()
```

```python
# ============================================================
# 5.1 INTRODUCE CONTROLLED DATA QUALITY ISSUES
# ============================================================

# I am intentionally introducing data quality issues
# to simulate real-world conditions.
#
# This allows me to practice investigation techniques
# that apply to any dataset, real or synthetic.
#
# These issues are placed at specific indices so I can
# verify later that my investigation techniques found them.
# ============================================================

print("=" * 60)
print("SECTION 5: DATA QUALITY INVESTIGATION")
print("=" * 60)
print()
print("Step 1: Introducing controlled data quality issues")
print("-" * 60)
print()

# ---- Issue 1: Missing satisfaction value ----
customers[100]["Service_Satisfaction"] = None
print("ISSUE 1: Missing Value")
print(f"   - Customer: {customers[100]['Customer_ID']}")
print(f"   - Variable: Service_Satisfaction")
print(f"   - Value: None (missing)")
print(f"   - Expected: 1 to 5")
print()

# ---- Issue 2: Negative monthly charge ----
customers[250]["Monthly_Charges"] = -5000
print("ISSUE 2: Negative Value")
print(f"   - Customer: {customers[250]['Customer_ID']}")
print(f"   - Variable: Monthly_Charges")
print(f"   - Value: {customers[250]['Monthly_Charges']}")
print(f"   - Expected: >= 0")
print()

# ---- Issue 3: Negative data usage ----
customers[400]["Data_Usage_GB"] = -10
print("ISSUE 3: Negative Value")
print(f"   - Customer: {customers[400]['Customer_ID']}")
print(f"   - Variable: Data_Usage_GB")
print(f"   - Value: {customers[400]['Data_Usage_GB']} GB")
print(f"   - Expected: >= 0")
print()

# ---- Issue 4: Impossible age ----
customers[600]["Age"] = 12
print("ISSUE 4: Value Outside Expected Range")
print(f"   - Customer: {customers[600]['Customer_ID']}")
print(f"   - Variable: Age")
print(f"   - Value: {customers[600]['Age']} years")
print(f"   - Expected: 18 to 100 years")
print()

# ---- Issue 5: Invalid satisfaction score ----
customers[800]["Service_Satisfaction"] = 7
print("ISSUE 5: Value Outside Expected Range")
print(f"   - Customer: {customers[800]['Customer_ID']}")
print(f"   - Variable: Service_Satisfaction")
print(f"   - Value: {customers[800]['Service_Satisfaction']}")
print(f"   - Expected: 1 to 5")
print()

# ---- Issue 6: Invalid churn value ----
customers[1000]["Churn"] = 2
print("ISSUE 6: Invalid Binary Value")
print(f"   - Customer: {customers[1000]['Customer_ID']}")
print(f"   - Variable: Churn")
print(f"   - Value: {customers[1000]['Churn']}")
print(f"   - Expected: 0 or 1")
print()

# ---- Issue 7: Duplicate Customer ID ----
customers[1500]["Customer_ID"] = customers[1499]["Customer_ID"]
print("ISSUE 7: Duplicate Record")
print(f"   - Customer 1: {customers[1499]['Customer_ID']}")
print(f"   - Customer 2: {customers[1500]['Customer_ID']}")
print(f"   - Problem: Both customers have the same ID")
print(f"   - Expected: Unique Customer_ID for each record")
print()

# ---- Issue 8: Invalid region ----
customers[2000]["Region"] = "Unknown Region"
print("ISSUE 8: Invalid Categorical Value")
print(f"   - Customer: {customers[2000]['Customer_ID']}")
print(f"   - Variable: Region")
print(f"   - Value: {customers[2000]['Region']}")
print(f"   - Expected: Lagos, Abuja, Rivers, Oyo, Kano, Enugu, Kaduna, Ogun")
print()

# ---- Issue 9: Negative tenure ----
customers[2500]["Tenure_Months"] = -5
print("ISSUE 9: Negative Value")
print(f"   - Customer: {customers[2500]['Customer_ID']}")
print(f"   - Variable: Tenure_Months")
print(f"   - Value: {customers[2500]['Tenure_Months']} months")
print(f"   - Expected: >= 0")
print()

# ---- Issue 10: Negative voice usage ----
customers[3000]["Voice_Minutes"] = -100
print("ISSUE 10: Negative Value")
print(f"   - Customer: {customers[3000]['Customer_ID']}")
print(f"   - Variable: Voice_Minutes")
print(f"   - Value: {customers[3000]['Voice_Minutes']} minutes")
print(f"   - Expected: >= 0")
print()

print("=" * 60)
print("All 10 data quality issues introduced.")
print("They will now be investigated.")
print("=" * 60)
```

```python
# ============================================================
# 5.2 CHECK FOR MISSING VALUES
# ============================================================

# Missing values are gaps in the data.
# They occur when a value was never recorded.
#
# I check every column to see if any value is None.
#
# Percentage missing helps me understand the scale of the issue.
# ============================================================

print("=" * 60)
print("Step 2: Checking for missing values")
print("-" * 60)
print()

missing_values = {}

# ---- Check every column ----
for column in customers[0]:
    missing_count = 0
    for customer in customers:
        if customer[column] is None:
            missing_count += 1
    missing_values[column] = missing_count

# ---- Display results ----
print("Column | Missing Count | Percentage")
print("-" * 45)

found_missing = False

for column, count in missing_values.items():
    if count > 0:
        percentage = (count / len(customers)) * 100
        print(f"{column:<20} | {count:>12} | {round(percentage, 2):>10}%")
        found_missing = True

if not found_missing:
    print("No missing values found.")
else:
    print()
    print("Status: Missing values identified.")

print()
```

```python
# ============================================================
# 5.3 CHECK FOR DUPLICATE CUSTOMER IDs
# ============================================================

# Each customer should have a unique Customer_ID.
# If two customers share the same ID, that is a data quality issue.
#
# I check by comparing every ID to every other ID.
# ============================================================

print("=" * 60)
print("Step 3: Checking for duplicate Customer IDs")
print("-" * 60)
print()

customer_ids = []

for customer in customers:
    customer_ids.append(customer["Customer_ID"])

duplicate_ids = []

for customer_id in customer_ids:
    if customer_ids.count(customer_id) > 1:
        if customer_id not in duplicate_ids:
            duplicate_ids.append(customer_id)

if len(duplicate_ids) == 0:
    print("Status: No duplicate Customer IDs found.")
else:
    print("Status: Duplicate Customer IDs found:")
    for customer_id in duplicate_ids:
        print(f"   - {customer_id}")

print()
```

```python
# ============================================================
# 5.4 CHECK FOR INVALID AGES
# ============================================================

# Reasonable age range: 18 to 100
#
# Any age below 18 or above 100 is a data quality issue.
# ============================================================

print("=" * 60)
print("Step 4: Checking for invalid ages")
print("-" * 60)
print()

invalid_ages = []

for customer in customers:
    age = customer["Age"]
    if age < 18 or age > 100:
        invalid_ages.append({
            "Customer_ID": customer["Customer_ID"],
            "Age": customer["Age"]
        })

print("Number of invalid ages:", len(invalid_ages))
print()

if len(invalid_ages) > 0:
    print("Customers with invalid ages:")
    for issue in invalid_ages[:10]:
        print(f"   - {issue['Customer_ID']}: {issue['Age']} years")

print()
```

```python
# ============================================================
# 5.5 CHECK FOR NEGATIVE CHARGES
# ============================================================

# A customer cannot have a negative bill.
#
# I check Monthly_Charges and Total_Charges.
# ============================================================

print("=" * 60)
print("Step 5: Checking for negative charges")
print("-" * 60)
print()

negative_monthly_charges = []
negative_total_charges = []

for customer in customers:
    if customer["Monthly_Charges"] < 0:
        negative_monthly_charges.append({
            "Customer_ID": customer["Customer_ID"],
            "Monthly_Charges": customer["Monthly_Charges"]
        })
    if customer["Total_Charges"] < 0:
        negative_total_charges.append({
            "Customer_ID": customer["Customer_ID"],
            "Total_Charges": customer["Total_Charges"]
        })

print("Negative monthly charges:", len(negative_monthly_charges))
print("Negative total charges:", len(negative_total_charges))
print()

if len(negative_monthly_charges) > 0:
    print("Customers with negative monthly charges:")
    for issue in negative_monthly_charges[:5]:
        print(f"   - {issue['Customer_ID']}: ₦{issue['Monthly_Charges']}")

print()
```

```python
# ============================================================
# 5.6 CHECK FOR NEGATIVE USAGE
# ============================================================

# Usage cannot be negative.
# I check Data_Usage_GB, Voice_Minutes, and SMS_Count.
# ============================================================

print("=" * 60)
print("Step 6: Checking for negative usage")
print("-" * 60)
print()

negative_data_usage = []
negative_voice_usage = []
negative_sms_usage = []

for customer in customers:
    if customer["Data_Usage_GB"] < 0:
        negative_data_usage.append(customer["Customer_ID"])
    if customer["Voice_Minutes"] < 0:
        negative_voice_usage.append(customer["Customer_ID"])
    if customer["SMS_Count"] < 0:
        negative_sms_usage.append(customer["Customer_ID"])

print("Negative data usage:", len(negative_data_usage))
print("Negative voice usage:", len(negative_voice_usage))
print("Negative SMS count:", len(negative_sms_usage))
print()

if len(negative_data_usage) > 0:
    print("Customers with negative data usage:", negative_data_usage[:5])
if len(negative_voice_usage) > 0:
    print("Customers with negative voice usage:", negative_voice_usage[:5])
if len(negative_sms_usage) > 0:
    print("Customers with negative SMS usage:", negative_sms_usage[:5])

print()
```

```python
# ============================================================
# 5.7 CHECK FOR INVALID TENURE
# ============================================================

# Tenure must be greater than or equal to 0.
#
# A customer cannot have negative tenure.
# ============================================================

print("=" * 60)
print("Step 7: Checking for invalid tenure")
print("-" * 60)
print()

invalid_tenure = []

for customer in customers:
    tenure = customer["Tenure_Months"]
    if tenure < 0:
        invalid_tenure.append({
            "Customer_ID": customer["Customer_ID"],
            "Tenure_Months": customer["Tenure_Months"]
        })

print("Number of invalid tenure values:", len(invalid_tenure))
print()

if len(invalid_tenure) > 0:
    print("Customers with invalid tenure:")
    for issue in invalid_tenure[:5]:
        print(f"   - {issue['Customer_ID']}: {issue['Tenure_Months']} months")

print()
```

```python
# ============================================================
# 5.8 CHECK FOR INVALID SATISFACTION SCORES
# ============================================================

# Satisfaction must be between 1 and 5.
#
# Any score outside this range is invalid.
# None values are handled separately (missing value issue).
# ============================================================

print("=" * 60)
print("Step 8: Checking for invalid satisfaction scores")
print("-" * 60)
print()

invalid_satisfaction = []

for customer in customers:
    satisfaction = customer["Service_Satisfaction"]
    if satisfaction is not None:
        if satisfaction < 1 or satisfaction > 5:
            invalid_satisfaction.append({
                "Customer_ID": customer["Customer_ID"],
                "Service_Satisfaction": customer["Service_Satisfaction"]
            })

print("Number of invalid satisfaction scores:", len(invalid_satisfaction))
print()

if len(invalid_satisfaction) > 0:
    print("Customers with invalid satisfaction scores:")
    for issue in invalid_satisfaction[:5]:
        print(f"   - {issue['Customer_ID']}: {issue['Service_Satisfaction']}")

print()
```

```python
# ============================================================
# 5.9 CHECK FOR INVALID CHURN VALUES
# ============================================================

# Churn must be 0 (stayed) or 1 (churned).
#
# Any other value is invalid.
# ============================================================

print("=" * 60)
print("Step 9: Checking for invalid churn values")
print("-" * 60)
print()

invalid_churn = []

for customer in customers:
    churn = customer["Churn"]
    if churn != 0 and churn != 1:
        invalid_churn.append({
            "Customer_ID": customer["Customer_ID"],
            "Churn": customer["Churn"]
        })

print("Number of invalid churn values:", len(invalid_churn))
print()

if len(invalid_churn) > 0:
    print("Customers with invalid churn values:")
    for issue in invalid_churn[:5]:
        print(f"   - {issue['Customer_ID']}: {issue['Churn']}")

print()
```

```python
# ============================================================
# 5.10 CHECK FOR INVALID CATEGORICAL VALUES
# ============================================================

# Each categorical column has a valid list of values.
# I check every record against these lists.
# ============================================================

print("=" * 60)
print("Step 10: Checking for invalid categorical values")
print("-" * 60)
print()

# ---- Valid lists ----
valid_genders = ["Male", "Female"]
valid_regions = ["Lagos", "Abuja", "Rivers", "Oyo", "Kano", "Enugu", "Kaduna", "Ogun"]
valid_subscription_types = ["Basic", "Standard", "Premium"]
valid_contract_types = ["Monthly", "Quarterly", "Annual"]
valid_payment_methods = ["Bank Transfer", "Card", "USSD", "Mobile Money"]
valid_device_types = ["Android", "iPhone", "Feature Phone"]

# ---- Check each ----
invalid_genders = []
invalid_regions = []
invalid_subscription_types = []
invalid_contract_types = []
invalid_payment_methods = []
invalid_device_types = []

for customer in customers:
    if customer["Gender"] not in valid_genders:
        invalid_genders.append(customer["Customer_ID"])
    if customer["Region"] not in valid_regions:
        invalid_regions.append(customer["Customer_ID"])
    if customer["Subscription_Type"] not in valid_subscription_types:
        invalid_subscription_types.append(customer["Customer_ID"])
    if customer["Contract_Type"] not in valid_contract_types:
        invalid_contract_types.append(customer["Customer_ID"])
    if customer["Payment_Method"] not in valid_payment_methods:
        invalid_payment_methods.append(customer["Customer_ID"])
    if customer["Device_Type"] not in valid_device_types:
        invalid_device_types.append(customer["Customer_ID"])

print("Invalid genders:", len(invalid_genders))
print("Invalid regions:", len(invalid_regions))
print("Invalid subscription types:", len(invalid_subscription_types))
print("Invalid contract types:", len(invalid_contract_types))
print("Invalid payment methods:", len(invalid_payment_methods))
print("Invalid device types:", len(invalid_device_types))
print()

if len(invalid_regions) > 0:
    print("Customers with invalid regions:", invalid_regions[:5])

print()
```

## SECTION 6: DATA CLEANING

```python
# ============================================================
# 6.1 FIX MISSING VALUES
# ============================================================

# Issue: One customer had a missing Service_Satisfaction value.
#
# My decision: Replace the missing value with the median.
#
# Why median?
# - Median is more robust to outliers than the mean
# - It keeps the distribution more realistic
# - It avoids inventing a value that skews analysis
# ============================================================

print("=" * 60)
print("SECTION 6: DATA CLEANING")
print("=" * 60)
print()
print("Step 1: Fixing missing values")
print("-" * 60)
print()

# ---- Collect non-missing satisfaction values ----
satisfaction_values = []

for customer in customers:
    if customer["Service_Satisfaction"] is not None:
        satisfaction_values.append(customer["Service_Satisfaction"])

# ---- Sort and find median ----
satisfaction_values.sort()

middle_index = len(satisfaction_values) // 2
median_satisfaction = satisfaction_values[middle_index]

print("Number of non-missing satisfaction values:", len(satisfaction_values))
print("Median satisfaction value:", median_satisfaction)
print()

# ---- Replace missing values with median ----
replaced_count = 0

for customer in customers:
    if customer["Service_Satisfaction"] is None:
        customer["Service_Satisfaction"] = median_satisfaction
        replaced_count += 1

print("Missing values replaced:", replaced_count)
print()

# ---- Verify ----
missing_after = 0

for customer in customers:
    if customer["Service_Satisfaction"] is None:
        missing_after += 1

print("Missing values remaining:", missing_after)
print()
```

```python
# ============================================================
# 6.2 REMOVE DUPLICATE RECORDS
# ============================================================

# Issue: Two customers shared the same Customer_ID (C01500).
#
# My decision: Keep the first occurrence, remove the duplicate.
#
# Why this approach?
# - Customer_ID should uniquely identify one customer
# - The duplicate is a data entry error
# - Removing one duplicate is safer than merging inconsistent records
# ============================================================

print("=" * 60)
print("Step 2: Removing duplicate records")
print("-" * 60)
print()

seen_ids = []
cleaned_customers = []
duplicates_removed = 0

for customer in customers:
    if customer["Customer_ID"] not in seen_ids:
        seen_ids.append(customer["Customer_ID"])
        cleaned_customers.append(customer)
    else:
        duplicates_removed += 1

# ---- Replace customers with the cleaned list ----
customers = cleaned_customers

print("Duplicates removed:", duplicates_removed)
print("Records after cleaning:", len(customers))
print()

# ---- Verify ----
customer_ids_after = []

for customer in customers:
    customer_ids_after.append(customer["Customer_ID"])

duplicates_after = []

for customer_id in customer_ids_after:
    if customer_ids_after.count(customer_id) > 1:
        if customer_id not in duplicates_after:
            duplicates_after.append(customer_id)

print("Duplicates remaining:", len(duplicates_after))
print()
```

```python
# ============================================================
# 6.3 FIX INVALID AGES
# ============================================================

# Issue: One customer had an age of 12.
#
# My decision: Cap to the minimum valid age (18).
#
# Why cap?
# - Removing the record loses information
# - Capping preserves the record while fixing the problem
# - Capping is a common approach when a value is plausible but wrong
# ============================================================

print("=" * 60)
print("Step 3: Fixing invalid ages")
print("-" * 60)
print()

ages_fixed = 0

for customer in customers:
    if customer["Age"] < 18:
        customer["Age"] = 18
        ages_fixed += 1
    elif customer["Age"] > 100:
        customer["Age"] = 100
        ages_fixed += 1

print("Invalid ages fixed:", ages_fixed)
print()

# ---- Verify ----
invalid_after = 0

for customer in customers:
    if customer["Age"] < 18 or customer["Age"] > 100:
        invalid_after += 1

print("Invalid ages remaining:", invalid_after)
print()
```

```python
# ============================================================
# 6.4 FIX NEGATIVE CHARGES
# ============================================================

# Issue: One customer had a negative Monthly_Charges value.
#
# My decision: Convert to absolute value.
#
# Why absolute value?
# - The value was likely recorded with the wrong sign
# - The magnitude is still useful (5000)
# - Converting to positive preserves the record
# ============================================================

print("=" * 60)
print("Step 4: Fixing negative charges")
print("-" * 60)
print()

charges_fixed = 0

for customer in customers:
    if customer["Monthly_Charges"] < 0:
        customer["Monthly_Charges"] = abs(customer["Monthly_Charges"])
        charges_fixed += 1
    if customer["Total_Charges"] < 0:
        customer["Total_Charges"] = abs(customer["Total_Charges"])
        charges_fixed += 1

print("Negative charges fixed:", charges_fixed)
print()

# ---- Verify ----
negative_after = 0

for customer in customers:
    if customer["Monthly_Charges"] < 0 or customer["Total_Charges"] < 0:
        negative_after += 1

print("Negative charges remaining:", negative_after)
print()
```

```python
# ============================================================
# 6.5 FIX NEGATIVE USAGE
# ============================================================

# Issue: Some customers had negative usage values.
#
# My decision: Set to 0.
#
# Why 0?
# - Usage cannot be negative
# - Setting to 0 is a conservative choice
# - It preserves the record while fixing the problem
# - It reflects that no usage was recorded for that period
# ============================================================

print("=" * 60)
print("Step 5: Fixing negative usage")
print("-" * 60)
print()

usage_fixed = 0

for customer in customers:
    if customer["Data_Usage_GB"] < 0:
        customer["Data_Usage_GB"] = 0
        usage_fixed += 1
    if customer["Voice_Minutes"] < 0:
        customer["Voice_Minutes"] = 0
        usage_fixed += 1
    if customer["SMS_Count"] < 0:
        customer["SMS_Count"] = 0
        usage_fixed += 1

print("Negative usage values fixed:", usage_fixed)
print()

# ---- Verify ----
negative_after = 0

for customer in customers:
    if customer["Data_Usage_GB"] < 0:
        negative_after += 1
    if customer["Voice_Minutes"] < 0:
        negative_after += 1
    if customer["SMS_Count"] < 0:
        negative_after += 1

print("Negative usage values remaining:", negative_after)
print()
```

```python
# ============================================================
# 6.6 FIX INVALID TENURE
# ============================================================

# Issue: One customer had a negative Tenure_Months value.
#
# My decision: Set to 0.
#
# Why 0?
# - Tenure cannot be negative
# - Setting to 0 indicates the customer has not been with us
# - It preserves the record while fixing the problem
# ============================================================

print("=" * 60)
print("Step 6: Fixing invalid tenure")
print("-" * 60)
print()

tenure_fixed = 0

for customer in customers:
    if customer["Tenure_Months"] < 0:
        customer["Tenure_Months"] = 0
        tenure_fixed += 1

print("Invalid tenure values fixed:", tenure_fixed)
print()

# ---- Verify ----
invalid_after = 0

for customer in customers:
    if customer["Tenure_Months"] < 0:
        invalid_after += 1

print("Invalid tenure values remaining:", invalid_after)
print()
```

```python
# ============================================================
# 6.7 FIX INVALID SATISFACTION SCORES
# ============================================================

# Issue: One customer had a satisfaction score of 7 (outside range 1-5).
#
# My decision: Cap to the maximum valid value (5).
#
# Why cap?
# - Removing the record loses information
# - Capping preserves the record while fixing the problem
# - Capping is standard practice when a value is close but incorrect
# ============================================================

print("=" * 60)
print("Step 7: Fixing invalid satisfaction scores")
print("-" * 60)
print()

satisfaction_fixed = 0

for customer in customers:
    satisfaction = customer["Service_Satisfaction"]
    if satisfaction is not None:
        if satisfaction < 1:
            customer["Service_Satisfaction"] = 1
            satisfaction_fixed += 1
        elif satisfaction > 5:
            customer["Service_Satisfaction"] = 5
            satisfaction_fixed += 1

print("Invalid satisfaction scores fixed:", satisfaction_fixed)
print()

# ---- Verify ----
invalid_after = 0

for customer in customers:
    satisfaction = customer["Service_Satisfaction"]
    if satisfaction is not None:
        if satisfaction < 1 or satisfaction > 5:
            invalid_after += 1

print("Invalid satisfaction scores remaining:", invalid_after)
print()
```

```python
# ============================================================
# 6.8 FIX INVALID CHURN VALUES
# ============================================================

# Issue: One customer had a Churn value of 2 (should be 0 or 1).
#
# My decision: Set to 0 (conservative).
#
# Why 0?
# - 0 means "stayed" (the more common outcome)
# - I do not want to falsely label someone as churned
# - Conservative choices avoid inflating churn rate
# ============================================================

print("=" * 60)
print("Step 8: Fixing invalid churn values")
print("-" * 60)
print()

churn_fixed = 0

for customer in customers:
    if customer["Churn"] != 0 and customer["Churn"] != 1:
        customer["Churn"] = 0
        churn_fixed += 1

print("Invalid churn values fixed:", churn_fixed)
print()

# ---- Verify ----
invalid_after = 0

for customer in customers:
    if customer["Churn"] != 0 and customer["Churn"] != 1:
        invalid_after += 1

print("Invalid churn values remaining:", invalid_after)
print()
```

```python
# ============================================================
# 6.9 FIX INVALID REGIONS
# ============================================================

# Issue: One customer had "Unknown Region" as their region.
#
# My decision: Replace with the most common valid region.
#
# Why most common?
# - It preserves the distribution
# - It is more realistic than inventing a value
# - It does not bias the analysis toward any minority region
# ============================================================

print("=" * 60)
print("Step 9: Fixing invalid regions")
print("-" * 60)
print()

# ---- Find the most common region ----
valid_regions = ["Lagos", "Abuja", "Rivers", "Oyo", "Kano", "Enugu", "Kaduna", "Ogun"]

region_counts = {}

for region in valid_regions:
    region_counts[region] = 0

for customer in customers:
    if customer["Region"] in valid_regions:
        region_counts[customer["Region"]] += 1

most_common_region = valid_regions[0]
highest_count = 0

for region, count in region_counts.items():
    if count > highest_count:
        highest_count = count
        most_common_region = region

print("Most common region:", most_common_region)
print("Count:", highest_count)
print()

# ---- Fix invalid regions ----
regions_fixed = 0

for customer in customers:
    if customer["Region"] not in valid_regions:
        customer["Region"] = most_common_region
        regions_fixed += 1

print("Invalid regions fixed:", regions_fixed)
print()

# ---- Verify ----
invalid_after = 0

for customer in customers:
    if customer["Region"] not in valid_regions:
        invalid_after += 1

print("Invalid regions remaining:", invalid_after)
print()
```

```python
# ============================================================
# 6.10 VERIFY CLEANED DATASET
# ============================================================

# Now I verify the cleaning worked.
#
# I re-run all the checks from Section 5.
#
# If any issues remain, I need to investigate further.
# ============================================================

print("=" * 60)
print("Step 10: Verifying cleaned dataset")
print("-" * 60)
print()

# ---- Total records ----
print("DATASET SIZE")
print("-" * 40)
print("Total records:", len(customers))
print("Total columns:", len(customers[0]))
print()

# ---- Missing values ----
missing_after_cleaning = 0

for customer in customers:
    for column in customer:
        if customer[column] is None:
            missing_after_cleaning += 1

print("MISSING VALUES")
print("-" * 40)
print("Missing values remaining:", missing_after_cleaning)
print()

# ---- Duplicates ----
customer_ids_after = []

for customer in customers:
    customer_ids_after.append(customer["Customer_ID"])

duplicates_after_cleaning = []

for customer_id in customer_ids_after:
    if customer_ids_after.count(customer_id) > 1:
        if customer_id not in duplicates_after_cleaning:
            duplicates_after_cleaning.append(customer_id)

print("DUPLICATES")
print("-" * 40)
print("Duplicate Customer IDs remaining:", len(duplicates_after_cleaning))
print()

# ---- Invalid ages ----
invalid_ages_after = 0

for customer in customers:
    if customer["Age"] < 18 or customer["Age"] > 100:
        invalid_ages_after += 1

print("INVALID AGES")
print("-" * 40)
print("Invalid ages remaining:", invalid_ages_after)
print()

# ---- Negative charges ----
negative_charges_after = 0

for customer in customers:
    if customer["Monthly_Charges"] < 0:
        negative_charges_after += 1
    if customer["Total_Charges"] < 0:
        negative_charges_after += 1

print("NEGATIVE CHARGES")
print("-" * 40)
print("Negative charges remaining:", negative_charges_after)
print()

# ---- Negative usage ----
negative_usage_after = 0

for customer in customers:
    if customer["Data_Usage_GB"] < 0:
        negative_usage_after += 1
    if customer["Voice_Minutes"] < 0:
        negative_usage_after += 1
    if customer["SMS_Count"] < 0:
        negative_usage_after += 1

print("NEGATIVE USAGE")
print("-" * 40)
print("Negative usage values remaining:", negative_usage_after)
print()

# ---- Invalid tenure ----
invalid_tenure_after = 0

for customer in customers:
    if customer["Tenure_Months"] < 0:
        invalid_tenure_after += 1

print("INVALID TENURE")
print("-" * 40)
print("Invalid tenure remaining:", invalid_tenure_after)
print()

# ---- Invalid satisfaction ----
invalid_satisfaction_after = 0

for customer in customers:
    satisfaction = customer["Service_Satisfaction"]
    if satisfaction is not None:
        if satisfaction < 1 or satisfaction > 5:
            invalid_satisfaction_after += 1

print("INVALID SATISFACTION")
print("-" * 40)
print("Invalid satisfaction remaining:", invalid_satisfaction_after)
print()

# ---- Invalid churn ----
invalid_churn_after = 0

for customer in customers:
    if customer["Churn"] != 0 and customer["Churn"] != 1:
        invalid_churn_after += 1

print("INVALID CHURN")
print("-" * 40)
print("Invalid churn remaining:", invalid_churn_after)
print()

# ---- Invalid regions ----
invalid_regions_after = 0

for customer in customers:
    if customer["Region"] not in valid_regions:
        invalid_regions_after += 1

print("INVALID REGIONS")
print("-" * 40)
print("Invalid regions remaining:", invalid_regions_after)
print()

print("=" * 60)
print("CLEANING VERIFICATION COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# 7.2 COUNT CHURNED CUSTOMERS
# ============================================================

# Now I want to know:
# How many customers have churned?
#
# Churn = 1 means the customer left.
#
# I use a loop to count every customer with Churn = 1.
# ============================================================

print("=" * 60)
print("Step 2: Counting churned customers")
print("-" * 60)
print()

churned_customers = 0
churned_ids = []

for customer in customers:
    if customer["Churn"] == 1:
        churned_customers += 1
        churned_ids.append(customer["Customer_ID"])

print("Churned Customers:", churned_customers)
print()

# ---- Show a sample ----

print("Sample churned customer IDs (first 10):")
for customer_id in churned_ids[:10]:
    print("  -", customer_id)

print("...")
print()

# ---- Business interpretation ----

print("This tells me the size of the churn problem.")
print()
```

## SECTION 8: CHURN ANALYSIS

### Why This Section Matters

In Section 7, I learned that **50% of customers churn**.

That is a critical number.

### My Analytical Approach

```python
# ============================================================
# 8.1 CHURN VS SATISFACTION
# ============================================================

# Question: Do churned customers have lower satisfaction?
#
# Satisfaction scale:
# 1 = Very Low
# 2 = Low
# 3 = Neutral
# 4 = High
# 5 = Very High
#
# I calculate the average satisfaction for each group.
# ============================================================

print("=" * 60)
print("SECTION 8: CHURN ANALYSIS")
print("=" * 60)
print()
print("Analysis 1: Churn vs Satisfaction")
print("-" * 60)
print()

# ---- Average satisfaction of churned customers ----
total_satisfaction_churned = 0
churned_count = 0

for customer in customers:
    if customer["Churn"] == 1:
        total_satisfaction_churned += customer["Service_Satisfaction"]
        churned_count += 1

avg_satisfaction_churned = total_satisfaction_churned / churned_count

# ---- Average satisfaction of retained customers ----
total_satisfaction_retained = 0
retained_count = 0

for customer in customers:
    if customer["Churn"] == 0:
        total_satisfaction_retained += customer["Service_Satisfaction"]
        retained_count += 1

avg_satisfaction_retained = total_satisfaction_retained / retained_count

# ---- Display results ----
print("Average Satisfaction:")
print("-" * 40)
print(f"Churned Customers:  {avg_satisfaction_churned:.2f}")
print(f"Retained Customers: {avg_satisfaction_retained:.2f}")
print()

# ---- Difference ----
difference = avg_satisfaction_retained - avg_satisfaction_churned

print(f"Difference: {difference:.2f} points")
print()

# ---- Interpretation ----
print("Observation:")
print("-" * 40)

if avg_satisfaction_churned < avg_satisfaction_retained:
    print("Churned customers had LOWER satisfaction than retained customers.")
    print()
    print("This suggests an association between satisfaction and churn.")
else:
    print("No clear difference in satisfaction between groups.")

print()
print("Note: This shows an association, not causation.")
print()
```

```python
# ============================================================
# 8.2 CHURN VS COMPLAINTS
# ============================================================

# Question: Do churned customers complain more?
#
# I calculate the average number of complaints for each group.
# ============================================================

print("=" * 60)
print("Analysis 2: Churn vs Complaints")
print("-" * 60)
print()

# ---- Average complaints of churned customers ----
total_complaints_churned = 0
churned_count = 0

for customer in customers:
    if customer["Churn"] == 1:
        total_complaints_churned += customer["Complaints_Count"]
        churned_count += 1

avg_complaints_churned = total_complaints_churned / churned_count

# ---- Average complaints of retained customers ----
total_complaints_retained = 0
retained_count = 0

for customer in customers:
    if customer["Churn"] == 0:
        total_complaints_retained += customer["Complaints_Count"]
        retained_count += 1

avg_complaints_retained = total_complaints_retained / retained_count

# ---- Display results ----
print("Average Complaints:")
print("-" * 40)
print(f"Churned Customers:  {avg_complaints_churned:.2f}")
print(f"Retained Customers: {avg_complaints_retained:.2f}")
print()

# ---- Difference ----
difference = avg_complaints_churned - avg_complaints_retained

print(f"Difference: {difference:.2f} complaints")
print()

# ---- Interpretation ----
print("Observation:")
print("-" * 40)

if avg_complaints_churned > avg_complaints_retained:
    print("Churned customers had MORE complaints than retained customers.")
    print()
    print("This suggests customer-service problems may be a factor.")
else:
    print("No clear difference in complaints between groups.")

print()
```

```python
# ============================================================
# 8.3 CHURN VS TENURE
# ============================================================

# Question: Are newer customers more likely to churn?
#
# Tenure = number of months the customer has been with the company.
#
# I compare average tenure for churned vs retained.
# ============================================================

print("=" * 60)
print("Analysis 3: Churn vs Tenure")
print("-" * 60)
print()

# ---- Average tenure of churned customers ----
total_tenure_churned = 0
churned_count = 0

for customer in customers:
    if customer["Churn"] == 1:
        total_tenure_churned += customer["Tenure_Months"]
        churned_count += 1

avg_tenure_churned = total_tenure_churned / churned_count

# ---- Average tenure of retained customers ----
total_tenure_retained = 0
retained_count = 0

for customer in customers:
    if customer["Churn"] == 0:
        total_tenure_retained += customer["Tenure_Months"]
        retained_count += 1

avg_tenure_retained = total_tenure_retained / retained_count

# ---- Display results ----
print("Average Tenure (months):")
print("-" * 40)
print(f"Churned Customers:  {avg_tenure_churned:.2f} months")
print(f"Retained Customers: {avg_tenure_retained:.2f} months")
print()

# ---- Difference ----
difference = avg_tenure_retained - avg_tenure_churned

print(f"Difference: {difference:.2f} months")
print()

# ---- Interpretation ----
print("Observation:")
print("-" * 40)

if avg_tenure_churned < avg_tenure_retained:
    print("Churned customers had SHORTER tenure than retained customers.")
    print()
    print("This suggests newer customers may be at higher risk of churning.")
else:
    print("No clear difference in tenure between groups.")

print()
```

```python
# ============================================================
# 8.4 CHURN VS LAST ACTIVITY
# ============================================================

# Question: Are churned customers more inactive?
#
# Last_Activity_Days = number of days since the customer last used the service.
#
# Higher value = longer inactivity.
# ============================================================

print("=" * 60)
print("Analysis 4: Churn vs Last Activity")
print("-" * 60)
print()

# ---- Average inactivity of churned customers ----
total_activity_churned = 0
churned_count = 0

for customer in customers:
    if customer["Churn"] == 1:
        total_activity_churned += customer["Last_Activity_Days"]
        churned_count += 1

avg_activity_churned = total_activity_churned / churned_count

# ---- Average inactivity of retained customers ----
total_activity_retained = 0
retained_count = 0

for customer in customers:
    if customer["Churn"] == 0:
        total_activity_retained += customer["Last_Activity_Days"]
        retained_count += 1

avg_activity_retained = total_activity_retained / retained_count

# ---- Display results ----
print("Average Inactivity (days since last activity):")
print("-" * 40)
print(f"Churned Customers:  {avg_activity_churned:.2f} days")
print(f"Retained Customers: {avg_activity_retained:.2f} days")
print()

# ---- Difference ----
difference = avg_activity_churned - avg_activity_retained

print(f"Difference: {difference:.2f} days")
print()

# ---- Interpretation ----
print("Observation:")
print("-" * 40)

if avg_activity_churned > avg_activity_retained:
    print("Churned customers had LONGER periods of inactivity.")
    print()
    print("This suggests inactivity may be a warning sign before churn.")
else:
    print("No clear difference in activity between groups.")

print()
```

```python
# ============================================================
# 8.5 CHURN BY SUBSCRIPTION TYPE
# ============================================================

# Question: Which subscription package has the highest churn?
#
# I group customers by Subscription_Type and calculate
# churn rate for each group.
# ============================================================

print("=" * 60)
print("Analysis 5: Churn by Subscription Type")
print("-" * 60)
print()

subscription_types = ["Basic", "Standard", "Premium"]

print("Churn Rate by Subscription Type:")
print("-" * 50)
print(f"{'Package':<15} {'Total':<10} {'Churned':<10} {'Churn Rate'}")
print("-" * 50)

for sub_type in subscription_types:
    total = 0
    churned = 0

    for customer in customers:
        if customer["Subscription_Type"] == sub_type:
            total += 1
            if customer["Churn"] == 1:
                churned += 1

    if total > 0:
        rate = (churned / total) * 100
        print(f"{sub_type:<15} {total:<10} {churned:<10} {rate:.2f}%")

print()
```

```python
# ============================================================
# 8.6 CHURN BY CONTRACT TYPE
# ============================================================

# Question: Does contract length affect churn?
#
# I compare churn rate across contract types.
# ============================================================

print("=" * 60)
print("Analysis 6: Churn by Contract Type")
print("-" * 60)
print()

contract_types = ["Monthly", "Quarterly", "Annual"]

print("Churn Rate by Contract Type:")
print("-" * 50)
print(f"{'Contract':<15} {'Total':<10} {'Churned':<10} {'Churn Rate'}")
print("-" * 50)

for contract in contract_types:
    total = 0
    churned = 0

    for customer in customers:
        if customer["Contract_Type"] == contract:
            total += 1
            if customer["Churn"] == 1:
                churned += 1

    if total > 0:
        rate = (churned / total) * 100
        print(f"{contract:<15} {total:<10} {churned:<10} {rate:.2f}%")

print()
```

```python
# ============================================================
# 9.1 BUILD THE HIGH RISK CLASSIFIER
# ============================================================

# I start with the simplest version:
# Identifying HIGH RISK customers only.
#
# HIGH RISK definition:
#   Satisfaction <= 2
#   AND
#   Complaints >= 3
#   AND
#   Last_Activity_Days >= 15
#
# All three conditions must be true.
# ============================================================

print("=" * 60)
print("SECTION 9: RISK LOGIC & SCORING")
print("=" * 60)
print()
print("Step 1: Building the High Risk Classifier")
print("-" * 60)
print()

high_risk_count = 0

print("HIGH RISK CUSTOMERS")
print("=" * 60)
print()

for customer in customers:
    if (customer["Service_Satisfaction"] <= 2 and
        customer["Complaints_Count"] >= 3 and
        customer["Last_Activity_Days"] >= 15):

        high_risk_count += 1

        # Show first 10 for review
        if high_risk_count <= 10:
            print(f"Customer: {customer['Customer_ID']}")
            print(f"  - Satisfaction: {customer['Service_Satisfaction']}")
            print(f"  - Complaints: {customer['Complaints_Count']}")
            print(f"  - Last Activity: {customer['Last_Activity_Days']} days")
            print(f"  - Risk Level: HIGH")
            print()

print("...")
print()
print("Total High Risk Customers:", high_risk_count)
print()
```

```python
# ============================================================
# 9.2 BUILD THE FULL RISK CLASSIFIER
# ============================================================

# Now I expand the logic to classify EVERY customer
# into one of three levels:
#
# HIGH:
#   Satisfaction <= 2 AND Complaints >= 3 AND Last_Activity >= 15
#
# MEDIUM:
#   Satisfaction <= 3 OR Complaints >= 2 OR Last_Activity >= 10
#
# LOW:
#   Everything else
# ============================================================

print("=" * 60)
print("Step 2: Building the Full Risk Classifier")
print("-" * 60)
print()

# ---- Counters ----
high_count = 0
medium_count = 0
low_count = 0

# ---- Customer ID storage ----
high_ids = []
medium_ids = []
low_ids = []

# ---- Classify every customer ----
for customer in customers:

    if (customer["Service_Satisfaction"] <= 2 and
        customer["Complaints_Count"] >= 3 and
        customer["Last_Activity_Days"] >= 15):

        risk = "HIGH"
        high_count += 1
        high_ids.append(customer["Customer_ID"])

    elif (customer["Service_Satisfaction"] <= 3 or
          customer["Complaints_Count"] >= 2 or
          customer["Last_Activity_Days"] >= 10):

        risk = "MEDIUM"
        medium_count += 1
        medium_ids.append(customer["Customer_ID"])

    else:

        risk = "LOW"
        low_count += 1
        low_ids.append(customer["Customer_ID"])

# ---- Display distribution ----
print("RISK DISTRIBUTION")
print("=" * 60)
print()

total = high_count + medium_count + low_count
high_percentage = (high_count / total) * 100
medium_percentage = (medium_count / total) * 100
low_percentage = (low_count / total) * 100

print(f"High Risk:   {high_count:>5} ({high_percentage:.2f}%)")
print(f"Medium Risk: {medium_count:>5} ({medium_percentage:.2f}%)")
print(f"Low Risk:    {low_count:>5} ({low_percentage:.2f}%)")
print()
print(f"Total Customers: {total}")
print()
print("=" * 60)
print()
```

```python
# ============================================================
# 9.3 BUILD THE RISK SCORING SYSTEM
# ============================================================

# Instead of just High/Medium/Low, I now assign a numeric score.
#
# This gives more granularity than categories alone.
#
# SCORING SYSTEM
# --------------
#
# Satisfaction:
#   Satisfaction <= 2  -> +3 points
#   Satisfaction == 3  -> +2 points
#   Satisfaction >= 4  -> +0 points
#
# Complaints:
#   Complaints >= 4    -> +3 points
#   Complaints 2-3     -> +2 points
#   Complaints < 2     -> +0 points
#
# Last Activity:
#   Last_Activity >= 20 -> +3 points
#   Last_Activity 10-19 -> +2 points
#   Last_Activity < 10  -> +0 points
#
# RISK LEVELS
# -----------
#   0-2   -> LOW
#   3-5   -> MEDIUM
#   6-9   -> HIGH
# ============================================================

print("=" * 60)
print("Step 3: Building the Risk Scoring System")
print("-" * 60)
print()
print("Scoring Rules:")
print("-" * 40)
print("Satisfaction:")
print("   <= 2  -> +3 points")
print("   == 3  -> +2 points")
print("   >= 4  -> +0 points")
print()
print("Complaints:")
print("   >= 4  -> +3 points")
print("   2-3   -> +2 points")
print("   < 2   -> +0 points")
print()
print("Last Activity:")
print("   >= 20 -> +3 points")
print("   10-19 -> +2 points")
print("   < 10  -> +0 points")
print()
print("Risk Levels:")
print("   0-2   -> LOW")
print("   3-5   -> MEDIUM")
print("   6-9   -> HIGH")
print()

# ---- Apply scoring to every customer ----
# I will store the score and level directly in the customer dictionary

for customer in customers:

    risk_score = 0

    # ---- Satisfaction points ----
    if customer["Service_Satisfaction"] <= 2:
        risk_score += 3
    elif customer["Service_Satisfaction"] == 3:
        risk_score += 2

    # ---- Complaints points ----
    if customer["Complaints_Count"] >= 4:
        risk_score += 3
    elif customer["Complaints_Count"] >= 2:
        risk_score += 2

    # ---- Last Activity points ----
    if customer["Last_Activity_Days"] >= 20:
        risk_score += 3
    elif customer["Last_Activity_Days"] >= 10:
        risk_score += 2

    # ---- Determine risk level ----
    if risk_score >= 6:
        risk_level = "High"
    elif risk_score >= 3:
        risk_level = "Medium"
    else:
        risk_level = "Low"

    # ---- Store results in the customer dictionary ----
    customer["Risk_Score"] = risk_score
    customer["Risk_Level"] = risk_level

print("Risk scoring applied to all customers.")
print()

# ---- Verify by showing first 5 customers ----
print("Sample Results (First 5 Customers):")
print("-" * 60)
print()

for i in range(5):
    customer = customers[i]
    print(f"Customer: {customer['Customer_ID']}")
    print(f"  Satisfaction: {customer['Service_Satisfaction']}")
    print(f"  Complaints: {customer['Complaints_Count']}")
    print(f"  Last Activity: {customer['Last_Activity_Days']} days")
    print(f"  Risk Score: {customer['Risk_Score']}")
    print(f"  Risk Level: {customer['Risk_Level']}")
    print()
```

```python
# ============================================================
# 10.1 CREATE TENURE GROUP
# ============================================================

# Tenure_Months tells me how long a customer has been with us.
#
# But a raw number is hard to interpret at a glance.
#
# So I create a "Tenure Group" that puts customers into
# lifecycle stages.
#
# CATEGORIES:
# - New Customer:       0-6 months
# - Early Customer:     7-12 months
# - Established:        13-24 months
# - Long-Term:          25+ months
#
# WHY THESE THRESHOLDS?
# - 6 months is the typical early engagement period
# - 12 months represents the first-year milestone
# - 24 months represents loyalty
# - Beyond 24 months is long-term loyalty
# ============================================================

print("=" * 60)
print("SECTION 10: FEATURE ENGINEERING")
print("=" * 60)
print()
print("Step 1: Creating Tenure Group")
print("-" * 60)
print()

print("Tenure Group Thresholds:")
print("-" * 40)
print("New Customer:       0-6 months")
print("Early Customer:     7-12 months")
print("Established:        13-24 months")
print("Long-Term:          25+ months")
print()

# ---- Create Tenure Group for every customer ----

for customer in customers:

    tenure = customer["Tenure_Months"]

    if tenure <= 6:
        group = "New Customer"
    elif tenure <= 12:
        group = "Early Customer"
    elif tenure <= 24:
        group = "Established"
    else:
        group = "Long-Term"

    customer["Tenure_Group"] = group

print("Tenure_Group added to all customers.")
print()

# ---- Show a sample ----

print("Sample Tenure Groups (First 10 Customers):")
print("-" * 60)
print()

for i in range(10):
    customer = customers[i]
    print(f"{customer['Customer_ID']}: Tenure = {customer['Tenure_Months']} months -> {customer['Tenure_Group']}")

print()
```

```python
# ============================================================
# 10.2 CREATE SPENDING GROUP
# ============================================================

# Monthly_Charges tells me how much a customer pays per month.
#
# By grouping customers into spending tiers, I can analyze
# whether spending level relates to churn.
#
# CATEGORIES:
# - Low Spend:      0-9,999
# - Medium Spend:   10,000-19,999
# - High Spend:     20,000+
#
# WHY THESE THRESHOLDS?
# - The distribution of monthly charges ranges from 2,000 to 30,000
# - These thresholds divide the range into three roughly equal bands
# - 10,000 and 20,000 are round, easy-to-explain numbers
# ============================================================

print("=" * 60)
print("Step 2: Creating Spending Group")
print("-" * 60)
print()

print("Spending Group Thresholds:")
print("-" * 40)
print("Low Spend:       0-9,999")
print("Medium Spend:    10,000-19,999")
print("High Spend:      20,000+")
print()

# ---- Create Spending Group for every customer ----

for customer in customers:

    charge = customer["Monthly_Charges"]

    if charge < 10000:
        group = "Low Spend"
    elif charge < 20000:
        group = "Medium Spend"
    else:
        group = "High Spend"

    customer["Spending_Group"] = group

print("Spending_Group added to all customers.")
print()

# ---- Show a sample ----

print("Sample Spending Groups (First 10 Customers):")
print("-" * 60)
print()

for i in range(10):
    customer = customers[i]
    print(f"{customer['Customer_ID']}: ₦{customer['Monthly_Charges']:>6} -> {customer['Spending_Group']}")

print()
```

```python
# ============================================================
# 10.3 CREATE ENGAGEMENT LEVEL
# ============================================================

# Engagement Level measures how actively a customer uses the service.
#
# I combine three usage variables:
# - Data_Usage_GB
# - Voice_Minutes
# - SMS_Count
#
# I create a simple engagement score and then classify customers
# as Low, Medium, or High engagement.
#
# ENGAGEMENT SCORE:
# - Data Usage:    +1 if >= 50 GB, +2 if >= 25 GB, +3 if < 25 GB
#   (Lower usage = higher engagement score, because low usage is a risk)
#
# Actually, let me reconsider. Higher usage means higher engagement.
# Let me redefine:
#
# - Data Usage:       +1 if >= 50 GB, +0 otherwise
# - Voice Minutes:    +1 if >= 2000 minutes, +0 otherwise
# - SMS Count:        +1 if >= 500 SMS, +0 otherwise
#
# SCORE RANGE: 0 to 3
#
# ENGAGEMENT LEVELS:
# - Low:     Score = 0
# - Medium:  Score = 1-2
# - High:    Score = 3
# ============================================================

print("=" * 60)
print("Step 3: Creating Engagement Level")
print("-" * 60)
print()

print("Engagement Level Rules:")
print("-" * 40)
print("+1 point if Data_Usage_GB >= 50 GB")
print("+1 point if Voice_Minutes >= 2000 minutes")
print("+1 point if SMS_Count >= 500 SMS")
print()
print("Score Range: 0 to 3")
print()
print("Engagement Levels:")
print("-" * 40)
print("Low:     Score = 0")
print("Medium:  Score = 1-2")
print("High:    Score = 3")
print()

# ---- Create Engagement Level for every customer ----

for customer in customers:

    score = 0

    if customer["Data_Usage_GB"] >= 50:
        score += 1
    if customer["Voice_Minutes"] >= 2000:
        score += 1
    if customer["SMS_Count"] >= 500:
        score += 1

    if score == 0:
        level = "Low"
    elif score <= 2:
        level = "Medium"
    else:
        level = "High"

    customer["Engagement_Score"] = score
    customer["Engagement_Level"] = level

print("Engagement_Level added to all customers.")
print()

# ---- Show a sample ----

print("Sample Engagement Levels (First 10 Customers):")
print("-" * 60)
print()

for i in range(10):
    customer = customers[i]
    print(f"{customer['Customer_ID']}: Data={customer['Data_Usage_GB']} GB, Voice={customer['Voice_Minutes']} min, SMS={customer['SMS_Count']} -> Score={customer['Engagement_Score']} -> {customer['Engagement_Level']}")

print()
```

```python
# ============================================================
# 10.4 CREATE SERVICE RISK
# ============================================================

# Service Risk measures service friction intensity.
#
# I combine three service-related variables:
# - Complaints_Count
# - Support_Calls
# - Network_Issues
#
# SERVICE RISK SCORE:
# - Complaints:    +2 if >= 4, +1 if >= 2
# - Support Calls: +2 if >= 7, +1 if >= 4
# - Network Issues:+2 if >= 5, +1 if >= 3
#
# SCORE RANGE: 0 to 6
#
# SERVICE RISK LEVELS:
# - Low:      Score = 0-1
# - Medium:   Score = 2-3
# - High:     Score = 4-6
# ============================================================

print("=" * 60)
print("Step 4: Creating Service Risk")
print("-" * 60)
print()

print("Service Risk Rules:")
print("-" * 40)
print("Complaints:")
print("   >= 4  -> +2 points")
print("   2-3   -> +1 point")
print()
print("Support Calls:")
print("   >= 7  -> +2 points")
print("   4-6   -> +1 point")
print()
print("Network Issues:")
print("   >= 5  -> +2 points")
print("   3-4   -> +1 point")
print()
print("Score Range: 0 to 6")
print()
print("Service Risk Levels:")
print("-" * 40)
print("Low:     Score = 0-1")
print("Medium:  Score = 2-3")
print("High:    Score = 4-6")
print()

# ---- Create Service Risk for every customer ----

for customer in customers:

    score = 0

    # Complaints
    if customer["Complaints_Count"] >= 4:
        score += 2
    elif customer["Complaints_Count"] >= 2:
        score += 1

    # Support Calls
    if customer["Support_Calls"] >= 7:
        score += 2
    elif customer["Support_Calls"] >= 4:
        score += 1

    # Network Issues
    if customer["Network_Issues"] >= 5:
        score += 2
    elif customer["Network_Issues"] >= 3:
        score += 1

    # Determine level
    if score <= 1:
        level = "Low"
    elif score <= 3:
        level = "Medium"
    else:
        level = "High"

    customer["Service_Risk_Score"] = score
    customer["Service_Risk_Level"] = level

print("Service_Risk_Level added to all customers.")
print()

# ---- Show a sample ----

print("Sample Service Risk (First 10 Customers):")
print("-" * 60)
print()

for i in range(10):
    customer = customers[i]
    print(f"{customer['Customer_ID']}: C={customer['Complaints_Count']}, S={customer['Support_Calls']}, N={customer['Network_Issues']} -> Score={customer['Service_Risk_Score']} -> {customer['Service_Risk_Level']}")

print()
```
